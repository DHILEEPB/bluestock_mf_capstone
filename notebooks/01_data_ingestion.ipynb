```python
# Imports
import pandas as pd
from pathlib import Path

# Define Paths
PROJECT_ROOT = Path().resolve().parent
RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print(f"Raw data directory: {RAW_DIR}")
print(f"Processed data directory: {PROCESSED_DIR}")
```

### Ingest Scheme Master Data

```python
if (RAW_DIR / "nav_history.csv").exists():
    df_nav = pd.read_csv(RAW_DIR / "nav_history.csv", nrows=100)
    # Standardize column headers
    df_nav.columns = [c.strip().lower().replace(" ", "_") for c in df_nav.columns]
    print("Columns standardized:", list(df_nav.columns))
    print(df_nav.head(3))
else:
    print("Demo file nav_history.csv not found in raw/. Run etl_pipeline.py first.")
```
